```python
 import json
  from pathlib import Path
  import matplotlib.pyplot as plt
  import numpy as np

  metrics_path = Path("Output/edbt/mobilitynebula_q2_q5_metrics.json")
  data = json.loads(metrics_path.read_text())

  qres = data["queries"]

  # Extract p50/p95 window_wait, processing, sink
  points = []
  labels = ["Q2 p50", "Q2 p95", "Q5 p50", "Q5 p95"]
  for qname, p in [("Q2", "p50"), ("Q2", "p95"), ("Q5", "p50"), ("Q5", "p95")]:
      m = qres[qname]["metrics"]["latency_ms"]
      points.append({
          "label": f"{qname} {p}",
          "window": m[f"{p}_window_wait"],
          "processing": m[f"{p}_processing"],
          "sink": m[f"{p}_sink"],
      })

  x = np.arange(len(points))
  width = 0.6

  window_vals = [p["window"] for p in points]
  proc_vals = [p["processing"] for p in points]
  sink_vals = [p["sink"] for p in points]

  fig, ax = plt.subplots(figsize=(6, 4))

  ax.bar(x, window_vals, width, label="Window wait")
  ax.bar(x, proc_vals, width, bottom=window_vals, label="Processing")
  bottom2 = np.array(window_vals) + np.array(proc_vals)
  ax.bar(x, sink_vals, width, bottom=bottom2, label="Sink")

  ax.set_xticks(x)
  ax.set_xticklabels(labels, rotation=30, ha="right")
  ax.set_ylabel("Latency (ms)")
  ax.set_title("Q2/Q5 p50/p95 latency breakdown (MobilityNebula)")
  ax.legend()

  plt.tight_layout()
  plt.show()
```
